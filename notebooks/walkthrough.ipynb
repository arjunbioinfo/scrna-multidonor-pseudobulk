# Multi-Donor scRNA-seq Integration & Pseudobulk PyDESeq2 Walkthrough

Inspects outputs from the Snakemake workflow (`make ci` or `make run`) on the Kang et al. (2018) 8-donor IFN-beta PBMC cohort (`GSE96583`).

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

RESULTS = Path('../results')
FIGURES = Path('../figures')

metrics = pd.read_csv(RESULTS / 'batch_metrics.tsv', sep='\t')
deg_summary = pd.read_csv(RESULTS / 'celltype_deg_summary.tsv', sep='\t')
pb_vs_sc = pd.read_csv(RESULTS / 'pseudobulk_vs_wilcoxon.tsv', sep='\t')
pathways = pd.read_csv(RESULTS / 'pathway_enrichment.tsv', sep='\t')

display(metrics)
display(deg_summary)
```

```python
# Compare pseudobulk PyDESeq2 vs naive cell-level Wilcoxon DEGs
display(pb_vs_sc)
display(pathways.sort_values('enrichment_z', ascending=False).head(10))
```

```python
# Top interferon-stimulated genes in the most responsive cell type
deseq = pd.read_csv(RESULTS / 'pseudobulk_deseq2_all.tsv', sep='\t')
top_ct = deg_summary.iloc[0]['cell_type']
deseq[(deseq['cell_type'] == top_ct) & deseq['significant']].sort_values('padj').head(15)
```
